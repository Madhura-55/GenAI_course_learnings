##Text Generation using Transformer model


```python
# Install required libraries
!pip install transformers torch -q

# Import pipeline (Hugging Face API)
from transformers import pipeline

# Load text generation pipeline
generator = pipeline(
    "text-generation",
    model="gpt2"
)
```

```python
# Input prompt
prompt = "Artificial Intelligence is"

# Generate text
generated_text = generator(
    prompt,
    max_new_tokens=40,
    do_sample=True,
    early_stopping=True,
    num_return_sequences=1
)

# Print output
print("Prompt:")
print(prompt)

print("\nGenerated Text:")
print(generated_text[0]['generated_text'])
```

gpt2 limitations: not trained to understand completion of a sentence, focused only on next token by token generation.
