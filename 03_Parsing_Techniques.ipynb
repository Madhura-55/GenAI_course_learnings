```python
# Install spaCy
!pip install spacy

# Import spaCy
import spacy

# Import visualization tool
from spacy import displacy

# Optional visualization support
import matplotlib.pyplot as plt

print("Libraries Installed Successfully")
```

```python
# Download English language model
!python -m spacy download en_core_web_sm

# Load the English NLP model
nlp = spacy.load("en_core_web_sm")

print("spaCy English Model Loaded Successfully")
```

```python
# Input sentence
sentence = "The quick brown fox jumps over the lazy dog."

# Process sentence using spaCy pipeline
doc = nlp(sentence)

print("Original Sentence:\n")
print(sentence)

print("\nDocument Processed Successfully")
```

```python
print(
    f"{'TOKEN':<15}"
    f"{'POS':<15}"
    f"{'DEPENDENCY':<15}"
    f"{'HEAD WORD'}"
)

print("-" * 60)

for token in doc:

    print(
        f"{token.text:<15}"
        f"{token.pos_:<15}"
        f"{token.dep_:<15}"
        f"{token.head.text}"
    )
```

```python
print("Dependency Relationships:\n")

for token in doc:

    print(
        f"{token.text}"
        f" --({token.dep_})--> "
        f"{token.head.text}"
    )
```

Label	Meaning
- det	determiner
- amod	adjective modifier
- nsubj	nominal subject
- prep	preposition
- pobj	object of preposition
- ROOT	main root verb

```python
# Visualize dependency parse tree
displacy.render(
    doc,
    style="dep",
    jupyter=True
)
```
