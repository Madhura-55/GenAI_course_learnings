###NLP preprocessing pipeline using NLTK:

- Tokenization
- Stopword Removal
- Stemming
- Lemmatization

```python
# Install NLTK
!pip install nltk

import nltk

# Download required resources
nltk.download('punkt')
nltk.download('punkt_tab')      # NEW (required in newer versions)
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
# Sample paragraph for NLP preprocessing

text = """
Natural Language Processing (NLP) is a field of Artificial Intelligence.
It helps computers understand human language.
I'm studying and analyzing text processing methods.
"""

# Display original text
print("Original Text:\n")
print(text)
```

```python
from nltk.tokenize import sent_tokenize, word_tokenize

# Sentence Tokenization
sentences = sent_tokenize(text)

print("Sentence Tokenization:\n")

for i, sentence in enumerate(sentences, start=1):
    print(f"Sentence {i}: {sentence}")

print("\n" + "="*50 + "\n")

# Word Tokenization
words = word_tokenize(text)

print("Word Tokenization:\n")
print(words)
```

```python
from nltk.corpus import stopwords

# Load English stopwords
stop_words = set(stopwords.words('english'))

print("Some stopwords:\n")
print(list(stop_words)[:20])

print("\n" + "="*50 + "\n")

# Remove stopwords from tokenized words
filtered_words = []

for word in words:

    # Convert to lowercase for comparison
    if word.lower() not in stop_words:
        filtered_words.append(word)

print("Words after Stopword Removal:\n")
print(filtered_words)
```

```python
from nltk.stem import PorterStemmer

# Create stemmer object
stemmer = PorterStemmer()

# Store stemmed words
stemmed_words = []

for word in filtered_words:

    stem_word = stemmer.stem(word)
    stemmed_words.append(stem_word)

print("Original Words:\n")
print(filtered_words)

print("\n" + "="*50 + "\n")

print("Stemmed Words:\n")
print(stemmed_words)
```

```python
from nltk.stem import WordNetLemmatizer

# Create lemmatizer object
lemmatizer = WordNetLemmatizer()

# Store lemmatized words
lemmatized_words = []

for word in filtered_words:

    lemma_word = lemmatizer.lemmatize(word)
    lemmatized_words.append(lemma_word)

print("Original Words:\n")
print(filtered_words)

print("\n" + "="*50 + "\n")

print("Lemmatized Words:\n")
print(lemmatized_words)
```

```python
print("="*60)
print("ORIGINAL TEXT")
print("="*60)

print(text)

print("\n" + "="*60)
print("TOKENIZED WORDS")
print("="*60)

print(words)

print("\n" + "="*60)
print("AFTER STOPWORD REMOVAL")
print("="*60)

print(filtered_words)

print("\n" + "="*60)
print("AFTER STEMMING")
print("="*60)

print(stemmed_words)

print("\n" + "="*60)
print("AFTER LEMMATIZATION")
print("="*60)

print(lemmatized_words)
```
